Step - 1 
Understanding and Exploring the DataSet 

```python
import pandas as pd

House_data = pd.read_csv("C:\Machine Learning\Projects\house-price-prediction-linear-regression\data\house_prices.csv")

print(House_data.head(10))
print(House_data.shape)
print(House_data.info())
```

We understood that the data contains 30 Rows and 6 Columns

Columns : 
            area_sqft
            bedrooms
            bathrooms
            age_years
            distance_city_km
            price_lakhs

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 3, figsize=(15, 5))

# 1. Area vs Price
ax[0].scatter(House_data["area_sqft"], House_data["price_lakhs"])
ax[0].set_xlabel("Area (sqft)")
ax[0].set_ylabel("Price (Lakhs)")
ax[0].set_title("Area vs Price")


# 2. Age vs Price
ax[1].scatter(House_data["age_years"], House_data["price_lakhs"])
ax[1].set_xlabel("Age (Years)")
ax[1].set_ylabel("Price (Lakhs)")
ax[1].set_title("Age vs Price")


# 3. Distance vs Price
ax[2].scatter(House_data["distance_city_km"], House_data["price_lakhs"])
ax[2].set_xlabel("Distance (km)")
ax[2].set_ylabel("Price (Lakhs)")
ax[2].set_title("Distance vs Price")


plt.tight_layout()
plt.show()
```

By the above graphs we understood 3 points 

1. Price increases if Area is increasing 
2. Price decreases if Age is increasing 
3. Price decreases if Distance is increasing
